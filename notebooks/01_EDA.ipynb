### Export Packages

```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
import numpy as np
```

### Read Data

We will first read the $R_{rs}$ data, and visualize it as Sentinel-3 OLCI spectra.

```python
df_rrs = pd.read_csv('./../data/rrs_chla_olci.csv', index_col=0)
df_rrs
```

```python
fig, ax = plt.subplots()
df_rrs_spectra = df_rrs.drop('Chla', axis=1)
df_rrs_spectra.T.plot(legend=False, ax=ax)
ax.set_xlabel("OLCI Bands (nm)")
ax.set_ylabel("$R_{rs}$ ($sr^{-1}$)")
ax.set_title(rf'$N={len(df_rrs_spectra)}$')
```

Now, we will statistically describe and visualize the $Chla$ column, which is the target variable containing the concentration of Chl-*a* pigment.

```python
df_rrs.describe()
```

```python
fig, ax = plt.subplots(1,2, figsize=(12,5))
sns.histplot(df_rrs['Chla'], kde=True, ax=ax[0])
sns.histplot(df_rrs['Chla'], kde=True, ax=ax[1], log_scale=True)
ax[0].set_xlabel('Chl-a [$mg/m^3$]')
ax[1].set_xlabel('Chl-a [$mg/m^3$]')
ax[0].set_title(rf'$N={len(df_rrs)}$' + '\nLinear Scale')
ax[1].set_title(rf'$N={len(df_rrs)}$' + '\nLog Scale')
```

Now, let us plot the correlation heatmap of each band, with $Chla$ concentration, to see how they are correlated.

```python
corr_matrix = df_rrs.corr()
mask = None
# mask = np.triu(np.ones_like(corr_matrix, dtype=bool)) # we are doing this to get rid of the upper triange of the matrix
```

```python
fig, ax = plt.subplots(figsize=(14, 10))
sns.heatmap(corr_matrix, annot=True, mask=mask, cmap='coolwarm', ax=ax, linewidths=0.5)
```
